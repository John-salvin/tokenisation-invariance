# 07 · What rescaling can fix (Section 7.1)

The failure has two parts.
* **Calibration fault (F1):** scores from different groups sit on different scales, so they cannot be pooled.
* **Sensitivity fault (F2):** the metric ranks translations wrongly within one group.

Example: if every romanised score sat a fixed amount below its native counterpart, pooling the two would rank a good romanised translation below a worse native one; that is F1. If a better romanised translation scores below a worse romanised one, that is F2.

Quantile normalisation (COMET-QN) maps each group's scores onto one reference distribution. It keeps the order inside every group, so it can fix F1 but never F2. This notebook checks both on the fragmentation buckets (about a minute).
