# 06 · How much meaning is actually lost? (Section 6)

A metric that scores romanised text lower is only at fault if the text is still as good a translation. Two checks.

1. **Information is also lost.** LaBSE, a sentence encoder not under test, finds the romanised translation of an English source less often.
2. **But the drop tracks fragmentation.** A near-lossless romanisation (ISO 15919) loses almost nothing, yet agreement drops just as much. Adding known amounts of character noise and extending the line to zero error measures the drop with nothing lost.
