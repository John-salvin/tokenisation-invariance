# 10 · Check every number in the paper

`paper_numbers.yaml` lists every number printed in the paper, where it is printed, and the table and rows it comes from. The check below recomputes each one from `results/tables/` and reports any mismatch. Run notebooks 01 to 09 first.

```python
import os
import sys
import warnings
from pathlib import Path

# quieten library notices that do not affect any result
os.environ["TRANSFORMERS_VERBOSITY"] = "error"      # "no PyTorch found": no model is needed here
warnings.filterwarnings("ignore", module="tqdm")
warnings.filterwarnings("ignore", module="transformers")

import pandas as pd

ROOT = Path.cwd().resolve().parent          # the repository root
sys.path.insert(0, str(ROOT / "src"))
TABLES = ROOT / "results" / "tables"
pd.set_option("display.width", 200)
pd.set_option("display.max_columns", 30)
pd.set_option("display.precision", 3)

def table(name):
    """A table written by the analysis, from results/tables/."""
    return pd.read_csv(TABLES / f"{name}.csv")
```

```python
import subprocess
r = subprocess.run([sys.executable, "scripts/verify_paper_numbers.py", "--log", "results/logs/verification.md"],
                   cwd=ROOT, capture_output=True, text=True)
print(r.stdout[-600:])
assert r.returncode == 0, "some numbers do not reproduce; see results/logs/verification.md"
```
