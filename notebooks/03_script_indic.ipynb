# 03 · Script Invariance on IndicMT Eval (Section 5.1)

**Question (RQ2).** When the same translation is written in another script, does a metric still agree with humans?

**What we did.** Eleven metrics scored every IndicMT Eval segment twice: in its native script and romanised with IndicXlit. The human score is the same for both. The metric scores are in `data/scores/metaeval_panel.parquet`; `experiments/script/` has the scoring code.

This notebook rebuilds Table 3 and the meta-evaluation tables (about 6 minutes).

```python
import os
import sys
import warnings
from pathlib import Path

# quieten library notices that do not affect any result
os.environ["TRANSFORMERS_VERBOSITY"] = "error"      # "no PyTorch found": no model is needed here
warnings.filterwarnings("ignore", module="tqdm")
warnings.filterwarnings("ignore", module="transformers")

import pandas as pd

ROOT = Path.cwd().resolve().parent          # the repository root
sys.path.insert(0, str(ROOT / "src"))
TABLES = ROOT / "results" / "tables"
pd.set_option("display.width", 200)
pd.set_option("display.max_columns", 30)
pd.set_option("display.precision", 3)

def table(name):
    """A table written by the analysis, from results/tables/."""
    return pd.read_csv(TABLES / f"{name}.csv")
```
