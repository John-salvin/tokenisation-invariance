# 04 · Script Invariance beyond Indic (Section 5)

**What we did.** Seven WMT settings in three non-Latin scripts were romanised with letter-for-letter schemes that can be reversed exactly (ISO 15919, ISO 9, an ISO 233-style mapping). Only translations that round-trip exactly, and that no metric would truncate, are compared. Serbian is the control: its Latin form uses slightly *fewer* tokens, so it changes the script without fragmenting the text.

This notebook rebuilds Table 3 and the Serbian control (about 10 minutes).
